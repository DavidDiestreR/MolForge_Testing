# Remove `SMILES_out` column from a CSV (overwrite in-place)
This notebook lets you pick **any CSV file path**, removes the column **`SMILES_out`** if present, and **overwrites the same file** keeping the same format (CSV).

⚠️ It will modify the file on disk.

```python
import os
from pathlib import Path
import pandas as pd

# --- 1) Set your CSV path here ---------------------------------
# You can use absolute or relative paths.
csv_path = Path(r"data/analysis_output/MolForge_MF_Analysis_2000_ECFP4/noise/noise0_0_noise1_1/Individual_Tanimoto_and_Canonical.csv")

print("CSV path:", csv_path)
assert csv_path.exists(), f"File not found: {csv_path}"
```

```python
# --- 2) Load, drop column if exists, overwrite -----------------
df = pd.read_csv(csv_path)

col_to_drop = "SMILES_out"
if col_to_drop in df.columns:
    df = df.drop(columns=[col_to_drop])
    print(f"✅ Dropped column: {col_to_drop}")
else:
    print(f"ℹ️ Column '{col_to_drop}' not found — nothing to drop.")

# Overwrite the same file
df.to_csv(csv_path, index=False)
print("✅ Overwritten:", csv_path)
print("Columns now:", list(df.columns))
```

## Notes
- If your CSV uses a different delimiter/encoding, change `pd.read_csv(...)` / `to_csv(...)` accordingly.
- By default it writes with comma delimiter and UTF-8 encoding.
